## 道路建设

```python
from math import inf


def daolu_pushe(n: int, s: str):
    # 表示对应下标有几个zore or one
    zero_arr = [0] * (n + 1)
    one_arr = [0] * (n + 1)
    for i in range(1, n + 1):
        zero_arr[i] = s[:i].count("0")
        one_arr[i] = s[n - i :].count("1")
    # print(zero_arr)
    # print(one_arr)
    min_diff = inf
    best_i = 0
    for i in range(n + 1):
        l_need = (i + 1) // 2
        r_need = (n - i + 1) // 2
        if zero_arr[i] >= l_need and one_arr[n - i] >= r_need:
            cur_diff = abs(n / 2 - i)
            if cur_diff < min_diff or (cur_diff == min_diff and i < best_i):
                min_diff, best_i = cur_diff, i
    print(best_i)


def main():
    n = int(input())
    for _ in range(n):
        a = int(input())
        b = input()
        daolu_pushe(a, b)


main()

# daolu_pushe(6, "011001")
```
