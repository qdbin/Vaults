```python
def main():
    """指定滑动窗口，求最大和"""
    m, n = map(int, input().split())
    arr = list(map(int, input().split()))
    ans = 0
    for l in range(len(arr) - n + 1):
        ans = max(ans, sum(arr[l : l + n]))
    print(ans)


main()
```

```python
from itertools import combinations


def solve():
    # 读取输入
    n, k = map(int, input().split())
    x_str = input().strip()
    y_str = input().strip()

    # 将字符串转换为数字列表，方便操作
    # digits_x[i] 和 digits_y[i] 分别表示 x 和 y 在第 i 位（从左到右）的数字
    digits_x = [int(c) for c in x_str]
    digits_y = [int(c) for c in y_str]

    min_diff = float('inf')

    # 必须保留的位数是 total_keep = n - k
    total_keep = n - k

    # 特殊情况：如果必须保留所有位 (k=0)，则保留位就是全部 n 位
    # 如果只保留 1 位，那只能是最后一位 (最低位)
    if total_keep == 0:
        print(0)
        return

    # 生成所有需要保留的位的索引组合
    # 最低位（索引 n-1）必须保留
    must_keep_index = n - 1
    other_indices = list(range(n - 1))  # 除了最低位之外的其他位索引

    # 我们需要从 other_indices 中选出 (total_keep - 1) 个索引，与 must_keep_index 一起构成保留位
    num_to_choose = total_keep - 1

    # 如果不需要再选其他位了 (num_to_choose == 0)
    if num_to_choose == 0:
        kept_indices_list = [[must_keep_index]]
    else:
        # 从 other_indices 中选择 num_to_choose 个
        kept_indices_list = list(combinations(other_indices, num_to_choose))
        # 将必须保留的最低位索引加入到每个组合的开头或结尾
        kept_indices_list = [list(comb) + [must_keep_index] for comb in kept_indices_list]

    # 遍历每一种保留位的组合
    for kept_indices in kept_indices_list:
        # 对保留的每一位，我们有两种分配方式 (swap or not)
        # 用二进制掩码 mask 来表示，0 表示不交换，1 表示交换
        num_kept = len(kept_indices)
        for mask in range(1 << num_kept):  # 从 0 到 2^num_kept - 1
            new_x_digits = []
            new_y_digits = []

            for i in range(num_kept):
                idx = kept_indices[i]  # 原始字符串中的索引
                bit = (mask >> i) & 1  # 当前位的分配选择

                if bit == 0:  # 不交换
                    new_x_digits.append(digits_x[idx])
                    new_y_digits.append(digits_y[idx])
                else:  # 交换
                    new_x_digits.append(digits_y[idx])
                    new_y_digits.append(digits_x[idx])

            # 将数字列表转换为整数
            # 例如 [9, 1, 0, 4] -> 9104
            x_prime = int(''.join(map(str, new_x_digits)))
            y_prime = int(''.join(map(str, new_y_digits)))

            diff = abs(x_prime - y_prime)
            if diff < min_diff:
                min_diff = diff

    print(min_diff)


# 运行主函数
solve()
```
